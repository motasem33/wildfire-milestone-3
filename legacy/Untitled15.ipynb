```python
import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

# Display all columns
pd.set_option("display.max_columns", None)

# Path to your dataset
file_path = r"C:\Users\New-pc\Desktop\wildfire_project\Algerian_forest_fires_dataset.csv"

# Load dataset
df = pd.read_csv(file_path)

df.head()
```

```python
# 1. تنظيف مسافةأسماءالأعمدة
df.columns = df.columns.str.strip()

# 2. تحديد المنطقتين
split_idx = df[df['day'].str.contains('Sidi-Bel Abbes', na=False)].index[0]
df.loc[:split_idx, 'Region'] = 0
df.loc[split_idx:, 'Region'] = 1
df['Region'] = df['Region'].astype(int)

# 3. حذف الصفوف النصية والترويسات المكررة
df = df.dropna(how='all')
df = df[~df['day'].str.contains('Sidi-Bel Abbes', na=False)]
df = df[df['day'].str.strip() != 'day']
df = df.reset_index(drop=True)

# 4. تنظيف عمود الفئات Classes
df['Classes'] = df['Classes'].str.strip()
df = df.dropna().reset_index(drop=True)

# 5. تصحيح أنواع البيانات
int_cols = ['day', 'month', 'year', 'Temperature', 'RH', 'Ws']
for col in int_cols:
    df[col] = df[col].astype(int)

float_cols = ['Rain', 'FFMC', 'DMC', 'DC', 'ISI', 'BUI', 'FWI']
for col in float_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df = df.dropna().reset_index(drop=True)


df.info()
```

```python
# Count missing values in each column
missing_count = df.isnull().sum()

# Calculate missing percentage
missing_percentage = (missing_count / len(df)) * 100

# Create a summary table
missing_summary = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": missing_count.values,
    "Missing Percentage (%)": missing_percentage.values,
    "Data Type": df.dtypes.astype(str).values
})

# Show columns that contain missing values first
missing_summary = missing_summary.sort_values(
    by="Missing Values",
    ascending=False
)

display(missing_summary.round(2))

print("Total missing cells:", int(df.isnull().sum().sum()))
print("Columns with missing values:", int((missing_count > 0).sum()))
```

```python
# Identify duplicated rows
duplicate_mask = df.duplicated()

# Count duplicated rows
duplicate_count = duplicate_mask.sum()

# Calculate percentage
duplicate_percentage = (
    duplicate_count / len(df) * 100
    if len(df) > 0 else 0
)

# Display summary
duplicate_analysis = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Duplicate Rows",
        "Duplicate Percentage (%)"
    ],
    "Value": [
        len(df),
        int(duplicate_count),
        round(duplicate_percentage, 2)
    ]
})

display(duplicate_analysis)

# Display examples of duplicate rows
if duplicate_count > 0:
    print("Examples of duplicated rows:")
    display(df[duplicate_mask].head(10))
else:
    print("No duplicate rows found.")
```
